## Installation

Dans un terminal placé dans le dossier `deep_learning_project`, exécuter :

```bash
pip install -r requirements.txt
pip install -r requirements-jupyter.txt
jupyter notebook
```

Puis ouvrir le dossier `notebooks/`.

## Paramètres conseillés pour ASUS VivoBook S14, 16 GB RAM

Pour éviter des temps de calcul trop longs :

- MLP : 30 epochs, batch size 32 ;
- CNN : 5 à 8 epochs, batch size 64, sous-ensemble Fashion-MNIST ;
- RNN/LSTM/GRU : 4 à 6 epochs, vocabulaire limité ;
- Seq2Seq : petit corpus intégré, 300 epochs.

Pour un rendu final plus solide, augmente les epochs lorsque tout fonctionne correctement.
