# Génération du rapport à partir des résultats

Après avoir exécuté les notebooks des parties I, II et III, ce notebook lit les résultats sauvegardés dans `results/` et génère un fichier de synthèse Markdown.

Le rapport Word/PDF fourni reste le rapport académique principal. Ce notebook sert surtout à produire automatiquement une annexe de résultats réels.

```python
from pathlib import Path
import os, sys, json

# Exécuter ce notebook depuis le dossier project ou depuis project/notebooks fonctionne.
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

print("Project root:", ROOT)
print("Python path ready")
```

```python
from pathlib import Path
import pandas as pd

result_files = [
    ROOT / "results" / "part1_mlp" / "summary_metrics.csv",
    ROOT / "results" / "part2_cnn" / "summary_metrics.csv",
    ROOT / "results" / "part3_rnn" / "summary_metrics.csv",
]

for path in result_files:
    print("
", path)
    if path.exists():
        display(pd.read_csv(path))
    else:
        print("Pas encore généré. Exécute d'abord le notebook correspondant.")
```
